```python
financial_statements = pl.read_parquet('../data/fmp_hist_financial_statements.parquet')#, columns=['date','symbol','RIC','cik','fiscalYear', 'period','cashAndCashEquivalents','averageInventory','minorityInterest','totalStockholdersEquity','totalAssets','shortTermDebt','longTermDebt','totalDebt','marketCap','commonStock','preferredStock','treasuryStock','financialLeverageRatio','growthEBITDA','netDebtToEBITDA'])

display(financial_statements.head(10))
```

```python
financial_statements = financial_statements.filter(pl.col("period")!='FY')

financial_statements = financial_statements.with_columns(pl.col("date").dt.cast_time_unit("us"))
```

```python
financial_statements = financial_statements.with_columns(((financial_statements.select(pl.col('marketCap','shortTermDebt','longTermDebt','minorityInterest','preferredStock')).sum_horizontal() - financial_statements.select(pl.col('cashAndCashEquivalents','inventory')).sum_horizontal())  / financial_statements['marketCap']).alias('MarketLeverage'))

financial_statements = financial_statements.with_columns(((financial_statements.select(pl.col('totalStockholdersEquity','shortTermDebt','longTermDebt','minorityInterest')).sum_horizontal() - financial_statements.select(pl.col('cashAndCashEquivalents','inventory')).sum_horizontal())  / (financial_statements['totalStockholdersEquity'] - financial_statements['preferredStock'])).alias('BookLeverage'))
```

## Merging all files to create Distance-to-Default and Spread-to-DD

**Definitions present in the literature: modeling-corporate-bond-returns page 55**

**Added paper on distance-to-default and spread-to-DD: Value Investing in Credit Markets**

```python
# Merge yieldbook with equity price data on Ticker/RIC and PricingDate/Date
equity_price_data_renamed = equity_price_data.select(pl.all().exclude(['Symbol','Date']).name.suffix("_eq"), 'Symbol', 'Date')
merged_df = yieldbook.join(equity_price_data_renamed, left_on=['Ticker', 'PricingDate'], right_on=['Symbol', 'Date'], how='left')
```

```python
merged_df = merged_df.sort(['RIC','PricingDate']).with_columns(pl.col("Close_eq").pct_change().over('RIC').alias("Ret_eq"))

merged_df = merged_df.sort(['RIC','PricingDate']).with_columns(pl.col('Ret_eq').rolling_std(window_size=12).over('RIC').alias('Vol_eq'))
```

```python
merged_df
```

```python
print(merged_df.shape)
merged_df.null_count()
```

```python
# financial_statements_renamed = financial_statements.select(pl.all().exclude(['symbol','date']).name.suffix("_fin"), 'symbol', 'date')
merged_df = merged_df.join(financial_statements.select(pl.col(['symbol','date','MarketLeverage','shortTermDebt','longTermDebt','totalDebt','marketCap'])), left_on=['Ticker', 'PricingDate'], right_on=['symbol', 'date'], how='left')
```

```python
merged_df = merged_df.sort(["RIC", "PricingDate"]).with_columns(pl.col(['MarketLeverage','shortTermDebt','longTermDebt','totalDebt','marketCap']).forward_fill().over("RIC"))
```

```python
merged_df = merged_df.with_columns((pl.col("Ret_eq") / pl.col('MarketLeverage')).alias("Ret_eq_unlevered"))
merged_df = merged_df.sort(['RIC','PricingDate']).with_columns(pl.col('Ret_eq').rolling_std(window_size=12).over('RIC').alias('Vol_eq_unlevered'))
```

```python
merged_df = merged_df.with_columns((pl.col('shortTermDebt') + 0.5*pl.col('longTermDebt')).alias('Default_Barrier'))

merged_df = merged_df.with_columns((pl.col('marketCap') + pl.col('totalDebt')).alias('Market_Value_Assets'))

merged_df = merged_df.with_columns((((pl.col('Market_Value_Assets') / pl.col('Default_Barrier')).log()) / (np.sqrt(12) * pl.col('Vol_eq_unlevered'))).alias('Distance_to_Default'))

merged_df = merged_df.with_columns((pl.col('OAS') / (1 - pl.col('Distance_to_Default').map_elements(norm.cdf))).alias('Spread_to_DD'))
```
