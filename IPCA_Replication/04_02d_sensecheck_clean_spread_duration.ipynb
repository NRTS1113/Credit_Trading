# Sense Check: Cleaned Spread & Duration Panel

This notebook inspects the cleaned dataset produced by `04_02c_clean_spread_duration.py`.
It reports schema, coverage, missingness, and a summary export for review.

```python
from pathlib import Path
import pandas as pd

# Use the notebook folder as the base directory
notebook_dir = Path.cwd()
# Outputs live under the local outputs folder
out_dir = notebook_dir / "outputs"

clean_path = out_dir / "bond_panel_monthly_clean_with_spread_duration_filtered.parquet"
if not clean_path.exists():
    raise FileNotFoundError(
        f"Missing {clean_path}. Run 04_02c_clean_spread_duration.py first."
    )

clean = pd.read_parquet(clean_path, engine="fastparquet")
clean.head()
```

```python
# Inspect column dtypes and non-null counts
clean.info()
```

```python
# Coverage: date range, unique bonds, and total rows
clean["month_end"].min(), clean["month_end"].max(), clean["RIC"].nunique(), len(clean)
```

```python
# Missingness rates (top 10)
clean.isna().mean().sort_values(ascending=False).head(10)
```

```python
# Return diagnostics (credit excess returns)
clean["credit_excess_return"].describe(percentiles=[0.01, 0.05, 0.5, 0.95, 0.99])
```
