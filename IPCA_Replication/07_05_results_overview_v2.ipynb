# Extended Sample Replication (2004+) — Not Pursued

The original plan for `07_xx` was to replicate the IPCA pipeline using only the 8 characteristics with full coverage from 2004 onwards (`bond_age`, `coupon`, `face_value`, `duration`, `spread`, `rating`, `mom_6m_equity`, `equity_volatility`), allowing a longer sample that spans the 2008 financial crisis.

**Why this was not pursued:**

The coverage heatmap shows that all 8 characteristics are available from 2004-07 with 100% coverage. However, the bond return data (`bond_panel_monthly_clean_with_spread_duration_filtered.parquet`) has extremely sparse coverage before 2009:

| Year | Bond-month observations |
|:----:|:-----------------------:|
| 2004 | 5 |
| 2006 | 10 |
| 2007 | 38 |
| 2008 | 147 |
| 2009 | 171 |
| 2011 | 554 |

The characteristics and bond returns come from different data sources — characteristics from FMP/Norgate (available from 2004), bond returns from the ICE-based panel (meaningful coverage only from ~2009 onwards). After merging the two, the pre-2009 months have too few bonds to support IPCA cross-section estimation, and are dropped by the `MIN_BONDS` filter.

**Conclusion:** The effective start date of our data is constrained by the bond return panel, not the characteristics. Both the 28-characteristic (`06_xx`) and 8-characteristic (`07_xx`) replications start from approximately the same period (2009-2011). Given this, there is no meaningful gain from reducing the characteristic set, and we retain the 28-characteristic specification as our main result.
